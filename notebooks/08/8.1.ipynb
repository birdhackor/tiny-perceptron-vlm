```python
examples = [
    {"answer": "4", "正確": True, "貼切比喻": False},
    {"answer": "4，就像兩雙筷子共有四根。", "正確": True, "貼切比喻": True},
    {"answer": "答案是5，數字如星光流動。", "正確": False, "貼切比喻": False},
]
for row in examples:
    print(row)
```
