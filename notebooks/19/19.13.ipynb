# 19.13 三個成品，分別拿來做什麼？


同一個小型客服工作台，可能同時需要三種工作：看商品或聽需求、追蹤多次修改後的狀態，以及判斷資料是否足夠。與其把所有輸出都叫「聊天」，這次讓三個成品各有清楚的輸入與驗收方式。

**MoE 助理**就是 19.1–19.12 的 V2 有限多模態助理，處理文字、商品圖、字卡（19.6 讀字用的中文字圖片）、客服語音與工具；權重與固定的測試成績沿用原版本，也保留尚未通過全部原定判準的結論。它的 Dense 版只供比較，不另算第四個主要成品。

**Looped 解題器**是一個 Dense 結構、反覆使用同一個區塊的有限文字解題模型。它讀固定語法的兩盒事件，例如 `兩盒追蹤;A=2;B=6;事件=A設5,A設1;查詢=A`，回答 A 的最後值 1；也處理三個 0–6 整數的加法。共用的區塊可以跑不同輪數，最後讀出一個答案類別。它從隨機權重學起，獨立載入與使用；它不是把 MoE 的輸出換個介面，也不宣稱一般聊天能力。兩種題目的輸入寫法見[19.15](https://birdhackor.github.io/tiny-perceptron-vlm/19.15.html)。

**機率判斷器**輸入狀態、問題與候選意義，回傳各候選的機率，可做服務選擇、資訊充分判斷及有限規準評級。它採用[9.11](https://birdhackor.github.io/tiny-perceptron-vlm/9.11.html)的 Jev 式輸入輸出契約，也就是事先約定輸入哪些欄位、輸出哪種機率；它有自己的權重與推論指令，就算不接 Looped 解題器，也能單獨判斷客服材料。它不冒充未公開的商業 Jev 配方。

| 成品 | 一個最小輸入 | 要檢查的輸出 |
| --- | --- | --- |
| MoE 助理 | 商品圖＋「右邊是什麼？」 | 依其有限類別與位置範圍生成回答 |
| Looped 解題器 | `兩盒追蹤;A=2;B=6;事件=A設5,A設1;查詢=A` | 答案類別 1；實際輪數 |
| 機率判斷器 | `卡片進度;申請編號=?`＋「必要資訊是否齊備？」 | yes／no 機率、選中候選與集中度 |

三個神經模型都用本專案訓練的權重，但任務與輸出形式不一樣。三個成品各有自己的推論指令，使用者依任務選擇；只有 Looped 解題器的輪數能對應到[7.23](https://birdhackor.github.io/tiny-perceptron-vlm/7.23.html)的思考投入（thinking effort）檔位，MoE 助理沒有因此得到原生的思考投入能力。後面的 Auto 是機率判斷器與 Looped 解題器合作的服務流程，不是第四個神經模型。

Looped 解題器與機率判斷器已從隨機權重完成訓練，19.15–19.18 使用實際測試題的結果，19.20 保留通過與失敗項目。閱讀不必自己重訓三個模型；短 Notebook 教機制，19.19 另提供固定權重的下載與使用方式。

練習：要問「有沒有足夠資訊查詢」應選哪個成品？機率判斷器；若已知有限語法事件、只要最後數值，則由 Looped 解題器處理。若輸入是任意照片，不能因 MoE 助理接受圖片就假定它辨識所有物件。
