# 19.20 什麼證據才足以把候選版叫成品？


一次示範成功，證明那個輸入在那個設定下能答對；訓練代價降低，證明模型更貼近受監督材料。要交付成品，還需要它在預先指定的新題上達到最小用途，且別人能用同一版本重做推論。

本輪在2026-10-16完成驗收。模型、資料SHA、溫度與Auto規則先提交到[測試前commit](https://github.com/birdhackor/tiny-perceptron-vlm/commit/821a2c717f05ad7b57e08e88cd965e92ff0c1200)，再首次執行測試；評分後沒有用這份測試資料改權重或選設定。兩位獨立工程審核者另從保存的原始分數，重算解題器、機制實驗、Auto與判斷器的結果。

| 事前要求 | 實際結果 | 判定 |
| --- | --- | --- |
| Looped 解題器：五輪追蹤至少90% | 2928/2956，99.05% | 主要門檻通過 |
| Looped 解題器：加法次要目標85% | 2/42，4.76% | 未達 |
| 機率判斷器：Choice至少90% | 11,797/11,797，100% | 主要門檻通過 |
| 機率判斷器：重排一致至少99% | 11,797/11,797，100% | 主要門檻通過 |

兩個新成品通過預定的主要門檻，加法失敗沒有被移除或降門檻。資訊充分、規準、校準、事件子群與分布變化分開報告；例如重複交換只有32/42，整體99.05%不能代替這個子群。判斷器的家族切分仍包含已披露的跨切分重複輸入句，100%不能改名成對一般中文或Jev所有問題的理解。

MoE 助理的固定測試紀錄已指出未達全部原定能力判準；新增兩個成品不會改寫這個結論。若新成品沒有達主要門檻，就標為候選或實驗，檢查錯誤、在授權預算內改善，不能因為三份檔案已上傳便宣布三成品完成。若資料規則或測試範圍要改，另開版本、說明理由，不在看完答案後悄悄降低門檻。

最後的實驗紀錄應接起四段證據：資料怎樣產生與隔離，模型怎樣訓練及選權重，凍結後如何評估，下載者能否實際使用。同時保存失敗例與限制，例如長度超界、錯誤但自信、改問句後失準，讓讀者知道何時應澄清或換工具。正式數字不以人工表格或手工狀態更新代替。

測試結果仍受樣本與訓練seed影響。剛好跨過門檻，不代表每一種實體、長度或問法都可靠；實驗紀錄須列實際樣本數、分任務結果與適用範圍。若在不同seed重訓，另外保存版本，不用最好一次覆蓋原始失敗。GPU正式訓練與CPU短練習各有目的，後者不代表完整訓練只能使用CPU。

閱讀這份教材時，可以先跑短小機制Notebook，再按[19.19](https://birdhackor.github.io/tiny-perceptron-vlm/19.19.html)下載固定版本測自己的有限輸入。完整的實驗紀錄 [final-report-v1.json](https://github.com/birdhackor/tiny-perceptron-vlm/blob/main/docs/lesson-products/results/final-report-v1.json) 包含1–6輪的結果（產品配方只監督1–5輪，第6輪看超出訓練深度時的表現）、分任務與子群、校準前後、變化測試及Auto成本；公開配套另附可重算成績的逐題原始分數。想研究架構才重做匹配實驗；想改服務才重做資料、校準與決策規則。三條路都從清楚的問題開始，沿著版本與證據判斷結果。

練習從實驗紀錄或[保存的錯誤例](https://github.com/birdhackor/tiny-perceptron-vlm/blob/main/docs/lesson-products/results/observed-failure-examples-v1.json)找一個失敗例：先按世界規則寫出正確答案，再檢查是哪一層出錯——資訊不足、候選規準、模型判斷、預算選擇，還是執行版本。只有找到層次，修正才對得到問題。
