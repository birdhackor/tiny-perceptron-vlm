# 19.20 什麼證據才足以把候選版叫成品？


一次示範成功，證明那個輸入在那個設定下能答對；訓練loss降低，證明模型更貼近受監督材料。要交付成品，還需要它在預先指定的新題上達到最小用途，且別人能用同一版本重做推論。

本輪B的核心要求是有限狀態追蹤最後測試至少90%正確；C的核心要求是Choice至少90%，且重排語義一致至少99%。題數、分母與逐題結果跟報告一起公開，不只給百分比。算術、資訊充分、等級、校準與分布變化各有自己的表，避免核心門檻掩蓋其他失敗。這些數字是開訓前的要求，實際成績尚待驗證。

A的固定報告已指出未達全部原定能力判準；新增B、C不會改寫這個歷史結論。若新產品沒有達核心門檻，就標為候選或實驗，檢查錯誤、在授權預算內改善，不能因為三份檔案已上傳便宣布三成品完成。若資料規則或測試範圍要改，另開版本、說明理由，不在看完答案後悄悄降低門檻。

最後報告應接起四段證據：資料怎樣產生與隔離，模型怎樣訓練及選權重，凍結後如何評估，下載者能否實際使用。同時保存失敗例與限制，例如長度超界、錯誤但自信、改問句後失準，讓讀者知道何時應澄清或換工具。正式數字不以人工表格或手工狀態更新代替。

最後測試的結果仍受樣本與訓練seed影響。剛好跨過門檻，不代表每一種實體、長度或問法都可靠；報告須列實際樣本數、分任務結果與適用範圍。若在不同seed重訓，另外保存版本，不用最好一次覆蓋原始失敗。GPU正式訓練與CPU短練習各有目的，後者不代表完整訓練只能使用CPU。

閱讀這份教材時，你可以先跑短小機制Notebook；待新增權重發布後，再下載三個成品測自己的有限範圍輸入。想研究架構才重做匹配實驗；想改服務才重做資料、校準與政策。三條路都從清楚的問題開始，沿著版本與證據判斷結果，而不是要求把所有技術一次套進同一個網路。

練習在新報告到位後找一個失敗例：先按世界規則寫出正解，再檢查是哪一層出錯——資訊不足、候選規準、模型判斷、預算選擇，還是執行版本。只有找到層次，修正才對得到問題。
