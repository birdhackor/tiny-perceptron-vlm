**最小實驗：**以下程式只處理這節的問題。


```python
candidates = [{"lr": 0.001, "validation": 0.7}, {"lr": 0.01, "validation": 0.8}]
selected = max(candidates, key=lambda r: r["validation"])
print("選擇", selected, "test仍封存；這裏只示範選擇規則")
```
